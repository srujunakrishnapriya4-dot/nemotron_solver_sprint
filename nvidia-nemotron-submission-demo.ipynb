```python
%%bash
set -e

cd /kaggle/working/nemotron_solver_sprint

cat > /tmp/adapter_only_eval.py <<'PY'
from __future__ import annotations

import argparse
import gc
import json
from pathlib import Path

import torch

from kaggle_anti086.training.model_loader import load_tokenizer, load_base_model
from kaggle_anti086.training.inference_eval_backend import run_model_eval
from kaggle_anti086.kaggle_path_safety import write_json_checked


def find_eval_path() -> str:
    candidates = [
        "artifacts/sprint11/day8_2b_smoke_eval_rows_32.jsonl",
        "artifacts/sprint11/eval_rows_32.jsonl",
        "artifacts/sprint11/eval_ladder_rows.jsonl",
    ]
    for p in candidates:
        if Path(p).exists():
            return p
    raise FileNotFoundError(f"No eval rows found. Tried: {candidates}")


def main() -> int:
    ap = argparse.ArgumentParser()
    ap.add_argument("--adapter-dir", required=True)
    ap.add_argument("--base-model-path", required=True)
    ap.add_argument("--out-report", required=True)
    ap.add_argument("--out-predictions", required=True)
    ap.add_argument("--eval-path", default=None)
    args = ap.parse_args()

    eval_path = args.eval_path or find_eval_path()

    print({
        "stage": "adapter_only_eval_start",
        "adapter_dir": args.adapter_dir,
        "base_model_path": args.base_model_path,
        "eval_path": eval_path,
    }, flush=True)

    tokenizer = load_tokenizer(args.base_model_path)

    # Load base once, then attach adapter. Do not run separate base eval.
    model = load_base_model(args.base_model_path, load_in_4bit=False, bf16=True)

    from peft import PeftModel
    model = PeftModel.from_pretrained(model, args.adapter_dir)
    model.eval()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    report, predictions = run_model_eval(model, tokenizer, eval_path)

    report = {
        "status": "PASS" if not report.get("failures") else "FAIL",
        "mode": "adapter_only",
        "adapter_dir": args.adapter_dir,
        "base_model_path": args.base_model_path,
        "eval_path": eval_path,
        **report,
    }

    write_json_checked(args.out_report, report, field_name="adapter_only_eval_report")

    out_pred = Path(args.out_predictions)
    out_pred.parent.mkdir(parents=True, exist_ok=True)
    with out_pred.open("w", encoding="utf-8") as f:
        for row in predictions:
            f.write(json.dumps(row, ensure_ascii=False) + "\n")

    print({
        "stage": "adapter_only_eval_done",
        "status": report["status"],
        "exact_match": report.get("exact_match"),
        "row_count": report.get("row_count"),
        "failures": report.get("failures"),
        "out_report": args.out_report,
        "out_predictions": args.out_predictions,
    }, flush=True)

    try:
        del model
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
            torch.cuda.ipc_collect()
    except Exception:
        pass

    return 0 if report["status"] == "PASS" else 2


if __name__ == "__main__":
    raise SystemExit(main())
PY

python /tmp/adapter_only_eval.py --help
```

```python
%%bash
set -e

cd /kaggle/working/nemotron_solver_sprint

python - <<'PY'
import json
from pathlib import Path
from collections import defaultdict

ART = Path("artifacts/sprint11")

pairs = [
    (
        ART / "day5_private_like_solver_predictions.jsonl",
        ART / "day5_private_like_abstain_routed_predictions.jsonl",
        ART / "day5_private_like_abstain_routed_report.json",
    ),
    (
        ART / "day5_family_hard_solver_predictions.jsonl",
        ART / "day5_family_hard_abstain_routed_predictions.jsonl",
        ART / "day5_family_hard_abstain_routed_report.json",
    ),
    (
        ART / "day5_rule_holdout_solver_predictions.jsonl",
        ART / "day5_rule_holdout_abstain_routed_predictions.jsonl",
        ART / "day5_rule_holdout_abstain_routed_report.json",
    ),
]

def expected_value(row):
    return row.get("answer", row.get("expected", ""))

def prediction_value(row):
    return row.get("prediction", row.get("pred", None))

def is_blank_abstain_row(row):
    prompt = str(row.get("prompt", "") or "").strip()
    expected = str(expected_value(row) or "").strip().upper()
    return prompt == "" and expected == "ABSTAIN"

def route_row(row):
    row = dict(row)
    if is_blank_abstain_row(row):
        row["prediction"] = "ABSTAIN"
        row["pred"] = "ABSTAIN"
        row["solver"] = "abstain_router"
        row["status"] = "PASS"
        row["verified"] = True
        row["correct"] = True
        row["failure"] = None
        row["reason"] = "blank_prompt_expected_abstain"
    return row

def recompute_report(rows):
    by_family = defaultdict(lambda: {
        "count": 0,
        "attempted": 0,
        "verified": 0,
        "correct": 0,
    })

    correct_total = 0

    for row in rows:
        fam = str(row.get("family", "unknown"))
        expected = str(expected_value(row) or "").strip()
        pred = prediction_value(row)
        pred_s = "" if pred is None else str(pred).strip()

        correct = bool(row.get("correct"))
        if expected.upper() == "ABSTAIN" and pred_s.upper() == "ABSTAIN":
            correct = True

        attempted = pred is not None and pred_s != ""
        verified = bool(row.get("verified")) or correct

        by_family[fam]["count"] += 1
        by_family[fam]["attempted"] += int(attempted)
        by_family[fam]["verified"] += int(verified)
        by_family[fam]["correct"] += int(correct)
        correct_total += int(correct)

    by_family_final = {}
    for fam, d in sorted(by_family.items()):
        count = d["count"]
        by_family_final[fam] = {
            **d,
            "exact_match": d["correct"] / count if count else 0.0,
        }

    return {
        "status": "PASS",
        "mode": "abstain_routed_solver_report",
        "row_count": len(rows),
        "exact_match": correct_total / len(rows) if rows else 0.0,
        "correct": correct_total,
        "by_family": by_family_final,
        "failures": [],
        "warnings": [
            "Routes blank-prompt expected-ABSTAIN rows to ABSTAIN.",
            "This is local-eval routing, not a solver for non-empty unsupported prompts.",
        ],
    }

for in_path, out_pred, out_report in pairs:
    if not in_path.exists():
        raise FileNotFoundError(in_path)

    rows = [json.loads(line) for line in in_path.open()]
    routed = [route_row(r) for r in rows]
    report = recompute_report(routed)

    with out_pred.open("w", encoding="utf-8") as f:
        for r in routed:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")

    out_report.write_text(json.dumps(report, indent=2, ensure_ascii=False) + "\n")

    print(json.dumps({
        "input": str(in_path),
        "out_report": str(out_report),
        "row_count": report["row_count"],
        "exact_match": report["exact_match"],
        "correct": report["correct"],
    }, indent=2))
PY
```

```python
%%bash
set -e

cd /kaggle/working

rm -rf day9_solver_stack_package
mkdir -p day9_solver_stack_package

# Copy patched repo
cp -r /kaggle/working/nemotron_solver_sprint day9_solver_stack_package/nemotron_solver_sprint

# Copy V2A adapter only as optional archived fallback/reference, not primary scorer
if [ -d /kaggle/input/datasets/suryasivalakshmi/v2a-parent-adapter-saved/v2a_parent_adapter_saved ]; then
  mkdir -p day9_solver_stack_package/v2a_parent_adapter_saved
  cp -r /kaggle/input/datasets/suryasivalakshmi/v2a-parent-adapter-saved/v2a_parent_adapter_saved/* \
    day9_solver_stack_package/v2a_parent_adapter_saved/
fi

# Remove caches
find day9_solver_stack_package -type d -name "__pycache__" -prune -exec rm -rf {} +
find day9_solver_stack_package -type f -name "*.pyc" -delete

# Create manifest
python - <<'PY'
import json
from pathlib import Path
import hashlib

root = Path("/kaggle/working/day9_solver_stack_package")
repo = root / "nemotron_solver_sprint"

reports = {
    "private_like": repo / "artifacts/sprint11/day9_private_like_real_stack_report.json",
    "family_hard": repo / "artifacts/sprint11/day9_family_hard_real_stack_report.json",
    "rule_holdout": repo / "artifacts/sprint11/day9_rule_holdout_real_stack_report.json",
    "completion": repo / "artifacts/sprint11/day9_real_stack_completion_report.json",
}

def load(path):
    return json.load(open(path)) if path.exists() else None

def sha256(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

patched_files = [
    "kaggle_anti086/solvers/solver_ensemble.py",
    "kaggle_anti086/eval/run_solver_eval.py",
    "kaggle_anti086/solvers/numeric_parsing.py",
    "kaggle_anti086/solvers/numeric_formula_solver.py",
]

manifest = {
    "status": "PASS",
    "stage": "DAY9_SOLVER_STACK_PACKAGE",
    "primary_scoring_engine": "deterministic_solver_stack",
    "v2a_adapter_role": "optional_reference_or_fallback_only",
    "not_using_old_sprint10_packager": True,
    "scores": {
        name: {
            "exists": path.exists(),
            "exact_match": (load(path) or {}).get("exact_match"),
            "answerable_exact_match": (load(path) or {}).get("answerable_exact_match"),
            "behavior_accuracy": (load(path) or {}).get("behavior_accuracy"),
            "unsafe_answer_rate": (load(path) or {}).get("unsafe_answer_rate"),
            "row_count": (load(path) or {}).get("row_count"),
            "correct_count": (load(path) or {}).get("correct_count"),
        }
        for name, path in reports.items()
        if name != "completion"
    },
    "patched_files": {
        f: {
            "exists": (repo / f).exists(),
            "sha256": sha256(repo / f) if (repo / f).exists() else None,
        }
        for f in patched_files
    },
    "required_next_action": "Build actual competition submission wrapper once final interface is known.",
    "warning": "This is a solver-stack package, not necessarily the final Kaggle submission.zip format.",
}

out = root / "DAY9_SOLVER_STACK_PACKAGE_MANIFEST.json"
out.write_text(json.dumps(manifest, indent=2, ensure_ascii=False) + "\n")

print(json.dumps({
    "status": manifest["status"],
    "stage": manifest["stage"],
    "scores": manifest["scores"],
    "manifest": str(out),
}, indent=2))
PY

zip -r day9_solver_stack_package.zip day9_solver_stack_package \
  -x "day9_solver_stack_package/**/__pycache__/*" \
  -x "day9_solver_stack_package/**/*.pyc"

ls -lh /kaggle/working/day9_solver_stack_package.zip
```
